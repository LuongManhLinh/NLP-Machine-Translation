```python
'''
Vietnamese-Lao Neural Machine Translation Model
'''

# ===== TASK 1: SETUP AND DEPENDENCIES =====
!pip install -q sacremoses subword-nmt sacrebleu sentencepiece pytorch-lightning
!pip install -q torchtext==0.18.0
!pip install -q nltk datasets
!pip install -q underthesea

import os
import re
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pad_sequence
import matplotlib.pyplot as plt
from collections import Counter, defaultdict
import time
import math
import json
import gc
from tqdm.notebook import tqdm
import sentencepiece as spm
from sacrebleu.metrics import BLEU
import nltk
nltk.download('punkt', quiet=True)
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from underthesea import sent_tokenize as vi_sent_tokenize
import warnings
warnings.filterwarnings('ignore')
from torch.optim.lr_scheduler import OneCycleLR

# Memory monitoring function
def show_gpu_memory():
    """Print current GPU memory usage"""
    if torch.cuda.is_available():
        print(f"GPU memory allocated: {torch.cuda.memory_allocated() / 1e9:.2f} GB")
        print(f"GPU memory reserved: {torch.cuda.memory_reserved() / 1e9:.2f} GB")
        print(f"GPU memory free: {torch.cuda.get_device_properties(0).total_memory - torch.cuda.memory_allocated():.2f} bytes")

# Set random seed for reproducibility
SEED = 123
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed(SEED)
torch.backends.cudnn.deterministic = True

# Check for GPU availability and set memory management options
if torch.cuda.is_available():
    # Enable memory efficient features for Colab
    torch.cuda.empty_cache()
    # Set memory allocation options to reduce fragmentation
    os.environ['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'
    print("GPU is available! Device:", torch.cuda.get_device_name(0))
    print(f"Total GPU memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
else:
    print("GPU not available, using CPU.")

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Set up mixed precision training - crucial for fitting large models in limited memory
use_amp = True if torch.cuda.is_available() else False
scaler = torch.cuda.amp.GradScaler() if use_amp else None
print(f"Using mixed precision training: {use_amp}")

# ===== TASK 2: DATA LOADING AND PREPROCESSING =====

from google.colab import drive
drive.mount('/content/drive')

def load_dataset(source_file, target_file):
    """Load parallel data from source and target files with improved processing."""
    with open(source_file, 'r', encoding='utf-8') as f:
        source_lines = [line.strip() for line in f if line.strip()]  # Skip empty lines

    with open(target_file, 'r', encoding='utf-8') as f:
        target_lines = [line.strip() for line in f if line.strip()]  # Skip empty lines

    # Handle length mismatch with warning
    if len(source_lines) != len(target_lines):
        min_len = min(len(source_lines), len(target_lines))
        print(f"WARNING: Source and target files have different lengths. Truncating to {min_len} lines.")
        source_lines = source_lines[:min_len]
        target_lines = target_lines[:min_len]

    # Basic cleaning
    cleaned_pairs = []
    for src, tgt in zip(source_lines, target_lines):
        # Filter out extremely long or short sentences
        if 1 <= len(src.split()) <= 500 and 1 <= len(tgt.split()) <= 500:
            # Normalize whitespace
            src = ' '.join(src.split())
            tgt = ' '.join(tgt.split())
            cleaned_pairs.append((src, tgt))

    print(f"Loaded {len(cleaned_pairs)} sentence pairs after cleaning")
    return cleaned_pairs

def lo_sent_tokenize(text):
    """
    Tách câu tiếng Lào đơn giản bằng regex (vì thiếu tokenizer chuyên biệt).
    """
    return [s.strip() for s in re.split(r'(?<=[.!?])\s+', text) if s.strip()]

def augment_data(data_pairs, lang_pair=('lo', 'vi')):
    """
    Tăng cường dữ liệu song ngữ bằng cách tách câu nếu dài.

    Args:
        data_pairs (List[Tuple[str, str]]): Cặp câu nguồn–đích.
        lang_pair (Tuple[str, str]): Ngôn ngữ của câu nguồn và đích (default là lo–vi).
    Returns:
        List[Tuple[str, str]]: Tập dữ liệu sau khi tăng cường.
    """
    augmented_pairs = []
    for src, tgt in data_pairs:
        augmented_pairs.append((src, tgt))  # Cặp gốc

        # Nếu cả 2 phía đều là câu dài, thực hiện phân tách
        if len(src.split()) > 15 and len(tgt.split()) > 15:
            # Chọn tokenizer theo ngôn ngữ
            if lang_pair[0] == 'vi':
                src_parts = vi_sent_tokenize(src)
            else:
                src_parts = lo_sent_tokenize(src)

            if lang_pair[1] == 'vi':
                tgt_parts = vi_sent_tokenize(tgt)
            else:
                tgt_parts = lo_sent_tokenize(tgt)

            # Chỉ chấp nhận nếu số câu tách ra bằng nhau và hợp lệ
            if len(src_parts) == len(tgt_parts) and len(src_parts) > 1:
                for s, t in zip(src_parts, tgt_parts):
                    if len(s.split()) >= 3 and len(t.split()) >= 3:
                        augmented_pairs.append((s, t))

    print(f"Augmented data: {len(augmented_pairs)} pairs (from {len(data_pairs)} original pairs)")
    return augmented_pairs

# Enhanced loading and preprocessing of data
base_dir = "/content/drive/MyDrive/VLSP2023"

# Train data
train_source_file = os.path.join(base_dir, "Train/train2023.lo")
train_target_file = os.path.join(base_dir, "Train/train2023.vi")
train_data = load_dataset(train_source_file, train_target_file)

# Dev data
dev_source_file = os.path.join(base_dir, "Dev/dev2023.lo")
dev_target_file = os.path.join(base_dir, "Dev/dev2023.vi")
dev_data = load_dataset(dev_source_file, dev_target_file)

# Public test data
test_source_file = os.path.join(base_dir, "Test/public/test_lo.txt")
test_target_file = os.path.join(base_dir, "Test/public/test_vi.txt")
test_data = load_dataset(test_source_file, test_target_file)

# Augmented train data with improved function
augmented_train_data = augment_data(train_data, lang_pair=('lo', 'vi'))
train_data = augmented_train_data  # Replace with augmented data

print(f"Train data size: {len(train_data)}")
print(f"Dev data size: {len(dev_data)}")
print(f"Test data size: {len(test_data)}")

# Reduce dataset size for memory-constrained environments
def sample_dataset(dataset, max_size):
    """Randomly sample a subset of the dataset"""
    if len(dataset) > max_size:
        return random.sample(dataset, max_size)
    return dataset

# Optionally reduce dataset size to fit in memory
# Comment this out if you have sufficient GPU memory
max_train_size = 100000  # Adjust based on your GPU memory
max_dev_size = 5000
max_test_size = 5000

train_data = sample_dataset(train_data, max_train_size)
dev_data = sample_dataset(dev_data, max_dev_size)
test_data = sample_dataset(test_data, max_test_size)

print(f"Reduced train data size: {len(train_data)}")
print(f"Reduced dev data size: {len(dev_data)}")
print(f"Reduced test data size: {len(test_data)}")

# ===== TASK 3: IMPROVED TOKENIZATION AND VOCABULARY BUILDING =====

def train_sentencepiece(data, lang, vocab_size=16000, model_type='bpe'):
    """Train a SentencePiece model with memory-efficient settings."""
    # Extract all text for the given language
    all_text = []
    for pair in data:
        if lang == 'lo':
            all_text.append(pair[0])
        else:  # lang == 'vi'
            all_text.append(pair[1])

    # Write sentences to a temporary file
    temp_file = f'temp_{lang}.txt'
    with open(temp_file, 'w', encoding='utf-8') as f:
        for text in all_text:
            f.write(text + '\n')

    # Configure SentencePiece training with memory-efficient settings
    spm.SentencePieceTrainer.train(
        input=temp_file,
        model_prefix=f'spm_{lang}',
        vocab_size=vocab_size,
        model_type=model_type,
        character_coverage=0.9995,
        pad_id=0,
        unk_id=1,
        bos_id=2,
        eos_id=3,
        normalization_rule_name='nmt_nfkc',
        num_threads=2,  # Reduce thread count for memory efficiency
        input_sentence_size=min(len(all_text), 100000),  # Limit training sentences
        split_by_unicode_script=True,
        split_by_whitespace=False
    )

    # Load the model
    sp = spm.SentencePieceProcessor()
    sp.load(f'spm_{lang}.model')

    return sp

# Train SentencePiece models with memory-efficient settings
lo_sp = train_sentencepiece(train_data, 'lo', vocab_size=16000)
vi_sp = train_sentencepiece(train_data, 'vi', vocab_size=16000)

print(f"Lao vocabulary size: {len(lo_sp)}")
print(f"Vietnamese vocabulary size: {len(vi_sp)}")

# Build vocabulary dictionaries
def build_vocab_from_sp(sp_model):
    """Build vocab dictionaries from SentencePiece model."""
    vocab_size = len(sp_model)

    # Special tokens
    PAD_IDX = 0
    UNK_IDX = 1
    BOS_IDX = 2
    EOS_IDX = 3

    # Create token to index and index to token mappings
    token_to_idx = {}
    idx_to_token = {}

    for i in range(vocab_size):
        token = sp_model.id_to_piece(i)
        token_to_idx[token] = i
        idx_to_token[i] = token

    return token_to_idx, idx_to_token, vocab_size

# Build vocabularies
lo_token_to_idx, lo_idx_to_token, lo_vocab_size = build_vocab_from_sp(lo_sp)
vi_token_to_idx, vi_idx_to_token, vi_vocab_size = build_vocab_from_sp(vi_sp)

# Constants for special tokens
PAD_IDX = 0
UNK_IDX = 1
BOS_IDX = 2
EOS_IDX = 3

# ===== TASK 4: ENHANCED DATASET AND DATALOADER CREATION =====

class TranslationDataset(Dataset):
    """Memory-efficient dataset for translation task."""
    def __init__(self, data, src_sp, tgt_sp, max_len=256):  # Reduced max_len for memory efficiency
        self.data = data
        self.src_sp = src_sp
        self.tgt_sp = tgt_sp
        self.max_len = max_len

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        src_text, tgt_text = self.data[idx]

        # Tokenize using SentencePiece
        src_tokens = [BOS_IDX] + self.src_sp.encode(src_text, out_type=int) + [EOS_IDX]
        tgt_tokens = [BOS_IDX] + self.tgt_sp.encode(tgt_text, out_type=int) + [EOS_IDX]

        # Truncate if necessary to avoid OOM issues
        if len(src_tokens) > self.max_len:
            src_tokens = src_tokens[:self.max_len-1] + [EOS_IDX]

        if len(tgt_tokens) > self.max_len:
            tgt_tokens = tgt_tokens[:self.max_len-1] + [EOS_IDX]

        return {
            'src': torch.tensor(src_tokens, dtype=torch.long),
            'tgt': torch.tensor(tgt_tokens, dtype=torch.long),
            'src_len': len(src_tokens),
            'tgt_len': len(tgt_tokens),
            'src_text': src_text,
            'tgt_text': tgt_text
        }

def collate_fn(batch):
    """Memory-efficient collate function."""
    # Sort batch by source sequence length (descending) for efficient packing
    batch = sorted(batch, key=lambda x: x['src_len'], reverse=True)

    # Get source and target sequences
    src_seqs = [item['src'] for item in batch]
    tgt_seqs = [item['tgt'] for item in batch]
    src_texts = [item['src_text'] for item in batch]
    tgt_texts = [item['tgt_text'] for item in batch]

    # Pad sequences
    src_padded = pad_sequence(src_seqs, batch_first=True, padding_value=PAD_IDX)
    tgt_padded = pad_sequence(tgt_seqs, batch_first=True, padding_value=PAD_IDX)

    # Prepare target input for teacher forcing (remove last token)
    tgt_input = tgt_padded[:, :-1].clone()

    # Prepare target output for loss calculation (remove first token)
    tgt_output = tgt_padded[:, 1:].clone()

    # Create padding masks
    src_key_padding_mask = (src_padded == PAD_IDX)
    tgt_key_padding_mask = (tgt_input == PAD_IDX)

    # Create target attention mask (to prevent looking at future tokens)
    tgt_len = tgt_input.size(1)
    tgt_attn_mask = torch.triu(
        torch.ones((tgt_len, tgt_len), dtype=torch.bool),
        diagonal=1
    )

    # Count target tokens for loss normalization
    ntokens = (tgt_output != PAD_IDX).sum().item()

    return {
        'src': src_padded,
        'tgt_input': tgt_input,
        'tgt_output': tgt_output,
        'src_key_padding_mask': src_key_padding_mask,
        'tgt_key_padding_mask': tgt_key_padding_mask,
        'tgt_attn_mask': tgt_attn_mask,
        'ntokens': ntokens,
        'src_texts': src_texts,
        'tgt_texts': tgt_texts
    }

# Create datasets with smaller max sequence length
train_dataset = TranslationDataset(train_data, lo_sp, vi_sp, max_len=256)
dev_dataset = TranslationDataset(dev_data, lo_sp, vi_sp, max_len=256)
test_dataset = TranslationDataset(test_data, lo_sp, vi_sp, max_len=256)

# Create dataloaders with smaller batch sizes for Google Colab
# Use a smaller batch size to fit in Colab's memory
BATCH_SIZE = 16  # Reduced from 32 or 64

# Create dataloaders
train_dataloader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_fn,
    num_workers=0,  # Reduced for Colab compatibility
    pin_memory=True if torch.cuda.is_available() else False
)

dev_dataloader = DataLoader(
    dev_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_fn,
    num_workers=0,
    pin_memory=True if torch.cuda.is_available() else False
)

test_dataloader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    collate_fn=collate_fn,
    num_workers=0,
    pin_memory=True if torch.cuda.is_available() else False
)

print(f"Number of training batches: {len(train_dataloader)}")
print(f"Number of validation batches: {len(dev_dataloader)}")
print(f"Number of test batches: {len(test_dataloader)}")

# ===== TASK 5: MEMORY-EFFICIENT TRANSFORMER MODEL =====

class PositionalEncoding(nn.Module):
    """Positional encoding for transformer model."""
    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super(PositionalEncoding, self).__init__()
        self.dropout = nn.Dropout(p=dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        """
        Args:
            x: Tensor, shape [batch_size, seq_len, embedding_dim]
        """
        x = x + self.pe[:, :x.size(1), :]
        return self.dropout(x)

class LabelSmoothedCrossEntropyLoss(nn.Module):
    """Cross entropy loss with label smoothing."""
    def __init__(self, smoothing=0.1, ignore_index=None):
        super(LabelSmoothedCrossEntropyLoss, self).__init__()
        self.smoothing = smoothing
        self.ignore_index = ignore_index
        self.confidence = 1.0 - smoothing

    def forward(self, pred, target):
        """
        Args:
            pred: Tensor of shape [batch, vocab_size]
            target: Tensor of shape [batch]
        """
        logprobs = torch.nn.functional.log_softmax(pred, dim=-1)
        nll_loss = -logprobs.gather(dim=-1, index=target.unsqueeze(1))
        nll_loss = nll_loss.squeeze(1)

        if self.ignore_index is not None:
            # Create a mask for non-ignored indices
            mask = (target != self.ignore_index).float()
            nll_loss = nll_loss * mask

            # Smooth loss only for non-ignored targets
            smooth_loss = -logprobs.sum(dim=-1) * mask

            # Normalize by the number of non-ignored elements
            non_ignored = mask.sum()
            if non_ignored > 0:
                nll_loss = nll_loss.sum() / non_ignored
                smooth_loss = smooth_loss.sum() / non_ignored
            else:
                return torch.tensor(0.0, device=pred.device)
        else:
            nll_loss = nll_loss.mean()
            smooth_loss = -logprobs.mean(dim=-1).mean()

        loss = self.confidence * nll_loss + self.smoothing * smooth_loss
        return loss

class MemoryEfficientTransformer(nn.Module):
    """Memory-efficient transformer for Colab with limited GPU."""
    def __init__(self, src_vocab_size, tgt_vocab_size,
                 d_model=512, nhead=8,
                 num_encoder_layers=6, num_decoder_layers=6,
                 dim_feedforward=2048, dropout=0.1,
                 activation="gelu"):
        super(MemoryEfficientTransformer, self).__init__()

        # Model dimensions
        self.d_model = d_model
        self.nhead = nhead

        # Embeddings
        self.src_embedding = nn.Embedding(src_vocab_size, d_model, padding_idx=PAD_IDX)
        self.tgt_embedding = nn.Embedding(tgt_vocab_size, d_model, padding_idx=PAD_IDX)
        self.positional_encoding = PositionalEncoding(d_model, dropout)

        # Enable checkpointing to save memory during backprop
        # While slower, this allows training larger models
        use_checkpoint = True

        # Create transformer layers
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            activation=activation,
            batch_first=True,
            norm_first=True  # Pre-LN Transformer (more stable)
        )

        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            activation=activation,
            batch_first=True,
            norm_first=True
        )

        # Create encoder and decoder
        self.encoder = nn.TransformerEncoder(
            encoder_layer=encoder_layer,
            num_layers=num_encoder_layers,
            norm=nn.LayerNorm(d_model)
        )

        self.decoder = nn.TransformerDecoder(
            decoder_layer=decoder_layer,
            num_layers=num_decoder_layers,
            norm=nn.LayerNorm(d_model)
        )

        # Output projection
        self.output_projection = nn.Linear(d_model, tgt_vocab_size)

        # Initialize parameters
        self._init_parameters()

        # Enable gradient checkpointing if available
        if use_checkpoint and hasattr(self.encoder, "layers"):
            for layer in self.encoder.layers:
                layer._checkpoint_gradients = True
            for layer in self.decoder.layers:
                layer._checkpoint_gradients = True

    def _init_parameters(self):
        """Initialize model parameters."""
        for p in self.parameters():
            if p.dim() > 1:
                nn.init.xavier_uniform_(p)

        # Special initialization for LayerNorm
        for name, p in self.named_parameters():
            if 'norm' in name and 'weight' in name:
                nn.init.constant_(p, 1.0)
            elif 'norm' in name and 'bias' in name:
                nn.init.constant_(p, 0.0)

    def forward(self, src, tgt_input, src_key_padding_mask=None, tgt_key_padding_mask=None, tgt_attn_mask=None):
        """Forward pass with memory optimization."""
        # Embed and positionally encode the source and target sequences
        src_embedded = self.positional_encoding(self.src_embedding(src) * math.sqrt(self.d_model))
        tgt_embedded = self.positional_encoding(self.tgt_embedding(tgt_input) * math.sqrt(self.d_model))

        # Pass through encoder
        memory = self.encoder(
            src_embedded,
            src_key_padding_mask=src_key_padding_mask
        )

        # Pass through decoder
        output = self.decoder(
            tgt_embedded,
            memory,
            tgt_mask=tgt_attn_mask,
            tgt_key_padding_mask=tgt_key_padding_mask,
            memory_key_padding_mask=src_key_padding_mask
        )

        # Project outputs to vocabulary size
        return self.output_projection(output)

    def generate(self, src, max_len=100, beam_size=3):
        """Generate translation with greedy search (memory-efficient)."""
        # Encode source sequence
        src_key_padding_mask = (src == PAD_IDX)

        # Embed source
        src_embedded = self.positional_encoding(self.src_embedding(src) * math.sqrt(self.d_model))

        # Encode
        memory = self.encoder(src_embedded, src_key_padding_mask=src_key_padding_mask)

        batch_size = src.size(0)
        device = src.device

        # Initialize with BOS tokens
        decoded_ids = torch.full((batch_size, 1), BOS_IDX, dtype=torch.long, device=device)

        # Simple greedy search for memory efficiency
        for i in range(max_len):
            # Create target mask
            tgt_len = decoded_ids.size(1)
            tgt_attn_mask = torch.triu(
                torch.ones((tgt_len, tgt_len), device=device, dtype=torch.bool),
                diagonal=1
            )

            # Embed target ids
            tgt_embedded = self.positional_encoding(self.tgt_embedding(decoded_ids) * math.sqrt(self.d_model))

            # Decode
            output = self.decoder(
                tgt_embedded,
                memory,
                tgt_mask=tgt_attn_mask,
                memory_key_padding_mask=src_key_padding_mask
            )

            # Get next token predictions
            logits = self.output_projection(output[:, -1])
            next_token = logits.argmax(dim=-1).unsqueeze(1)

            # Append to sequence
            decoded_ids = torch.cat([decoded_ids, next_token], dim=1)

            # Check if all sequences have EOS
            if (next_token == EOS_IDX).all():
                break

        return decoded_ids

# ===== TASK 6: TRAINING FUNCTION WITH MEMORY OPTIMIZATIONS =====

def count_parameters(model):
    """Count the number of trainable parameters in the model."""
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

# Initialize model with optimized parameters
d_model = 512
nhead = 8
num_encoder_layers = 6
num_decoder_layers = 6
dim_feedforward = 1024
dropout = 0.1

# Create model
model = MemoryEfficientTransformer(
    src_vocab_size=lo_vocab_size,
    tgt_vocab_size=vi_vocab_size,
    d_model=d_model,
    nhead=nhead,
    num_encoder_layers=num_encoder_layers,
    num_decoder_layers=num_decoder_layers,
    dim_feedforward=dim_feedforward,
    dropout=dropout,
    activation="gelu"
).to(device)

# Apply improved weight initialization
def init_weights(m):
    if isinstance(m, nn.Linear):
        nn.init.xavier_uniform_(m.weight, gain=1/math.sqrt(2))
        if m.bias is not None:
            nn.init.zeros_(m.bias)
    elif isinstance(m, nn.Embedding):
        nn.init.normal_(m.weight, mean=0, std=0.01)
    elif isinstance(m, nn.LayerNorm):
        nn.init.ones_(m.weight)
        nn.init.zeros_(m.bias)

model.apply(init_weights)

print(f"The model has {count_parameters(model):,} trainable parameters")

# Loss function (no label smoothing for simplicity and stability)
criterion = nn.CrossEntropyLoss(ignore_index=PAD_IDX)

# Optimizer with optimized learning rate
learning_rate = 1e-3
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=learning_rate,
    betas=(0.9, 0.98),
    eps=1e-8,
    weight_decay=0.01
)

# Advanced learning rate scheduler
scheduler = OneCycleLR(
    optimizer,
    max_lr=learning_rate,
    total_steps=len(train_dataloader) * 15,  # 15 epoch
    pct_start=0.1,  # Warm-up in first 10%
    div_factor=25,   # learning_rate_min = learning_rate/25
    final_div_factor=10000,
    anneal_strategy='cos'
)

# Gradient accumulation settings for simulating larger batch sizes
grad_accumulation_steps = 4  # Simulate a batch size 4x larger
effective_batch_size = BATCH_SIZE * grad_accumulation_steps
print(f"Using gradient accumulation with {grad_accumulation_steps} steps")
print(f"Effective batch size: {effective_batch_size}")

def train_epoch(model, dataloader, optimizer, criterion, grad_accumulation_steps=1,
                clip=1.0, use_amp=False, scaler=None, scheduler=None):
    """Train with gradient accumulation and mixed precision."""
    model.train()
    epoch_loss = 0

    # Clear cache before training
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    # Progress bar
    progress_bar = tqdm(dataloader, desc='Training')

    optimizer.zero_grad()

    for batch_idx, batch in enumerate(progress_bar):
        # Move tensors to device
        src = batch['src'].to(device)
        tgt_input = batch['tgt_input'].to(device)
        tgt_output = batch['tgt_output'].to(device)
        src_key_padding_mask = batch['src_key_padding_mask'].to(device)
        tgt_key_padding_mask = batch['tgt_key_padding_mask'].to(device)
        tgt_attn_mask = batch['tgt_attn_mask'].to(device)

        # Forward pass with mixed precision
        if use_amp:
            with torch.cuda.amp.autocast():
                output = model(
                    src=src,
                    tgt_input=tgt_input,
                    src_key_padding_mask=src_key_padding_mask,
                    tgt_key_padding_mask=tgt_key_padding_mask,
                    tgt_attn_mask=tgt_attn_mask
                )
                loss = criterion(output.reshape(-1, output.size(-1)), tgt_output.reshape(-1))
                # Scale loss for gradient accumulation
                loss = loss / grad_accumulation_steps

            # Backward pass with gradient scaling
            scaler.scale(loss).backward()

            # Step after accumulation
            if (batch_idx + 1) % grad_accumulation_steps == 0 or batch_idx == len(dataloader) - 1:
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), clip)
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad()

                # Step scheduler - update learning rate after each batch
                if scheduler is not None:
                    scheduler.step()
        else:
            output = model(
                src=src,
                tgt_input=tgt_input,
                src_key_padding_mask=src_key_padding_mask,
                tgt_key_padding_mask=tgt_key_padding_mask,
                tgt_attn_mask=tgt_attn_mask
            )
            loss = criterion(output.reshape(-1, output.size(-1)), tgt_output.reshape(-1))
            # Scale loss for gradient accumulation
            loss = loss / grad_accumulation_steps

            # Backward pass
            loss.backward()

            # Step after accumulation
            if (batch_idx + 1) % grad_accumulation_steps == 0 or batch_idx == len(dataloader) - 1:
                torch.nn.utils.clip_grad_norm_(model.parameters(), clip)
                optimizer.step()
                optimizer.zero_grad()

                # Step scheduler - update learning rate after each batch
                if scheduler is not None:
                    scheduler.step()

        # Record full-scale loss for reporting
        epoch_loss += loss.item() * grad_accumulation_steps

        # Display current learning rate
        current_lr = optimizer.param_groups[0]['lr']

        # Update progress bar
        progress_bar.set_postfix({
            'loss': epoch_loss / (batch_idx + 1),
            'lr': f"{current_lr:.2e}"
        })

        # Free memory
        del src, tgt_input, tgt_output, output, loss
        if batch_idx % 10 == 0:
            torch.cuda.empty_cache()

    # Garbage collection
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return epoch_loss / len(dataloader)

def evaluate(model, dataloader, criterion):
    """Evaluate model with memory optimizations."""
    model.eval()
    epoch_loss = 0

    # Clear cache before evaluation
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    with torch.no_grad():
        for batch in tqdm(dataloader, desc='Evaluating'):
            # Move tensors to device
            src = batch['src'].to(device)
            tgt_input = batch['tgt_input'].to(device)
            tgt_output = batch['tgt_output'].to(device)
            src_key_padding_mask = batch['src_key_padding_mask'].to(device)
            tgt_key_padding_mask = batch['tgt_key_padding_mask'].to(device)
            tgt_attn_mask = batch['tgt_attn_mask'].to(device)

            # Forward pass
            output = model(
                src=src,
                tgt_input=tgt_input,
                src_key_padding_mask=src_key_padding_mask,
                tgt_key_padding_mask=tgt_key_padding_mask,
                tgt_attn_mask=tgt_attn_mask
            )

            # Compute loss
            loss = criterion(output.reshape(-1, output.size(-1)), tgt_output.reshape(-1))
            epoch_loss += loss.item()

            # Free memory
            del src, tgt_input, tgt_output, output, loss

    # Garbage collection
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return epoch_loss / len(dataloader)

# ===== TASK 7: MEMORY-EFFICIENT BLEU CALCULATION =====

def calculate_bleu_sample(model, dataloader, lo_sp, vi_sp, device, max_samples=100):
    """Calculate BLEU score on a specified number of samples."""
    model.eval()
    bleu = BLEU()

    references = []
    hypotheses = []
    sample_count = 0

    with torch.no_grad():
        # Process batches until we reach max_samples
        for batch in tqdm(dataloader, desc="Calculating BLEU"):
            # Get source and reference text (use full batch)
            src = batch['src'].to(device)
            tgt_texts = batch['tgt_texts']

            # If adding this batch would exceed max_samples, only take what we need
            remaining = max_samples - sample_count
            if remaining <= 0:
                break

            if remaining < len(src):
                src = src[:remaining]
                tgt_texts = tgt_texts[:remaining]

            # Generate translations
            output_ids = model.generate(src, max_len=100, beam_size=5)

            # Convert tokens to text
            for i in range(len(output_ids)):
                # Get reference
                reference = tgt_texts[i]

                # Convert prediction
                pred_tokens = [t for t in output_ids[i].tolist() if t not in [PAD_IDX, BOS_IDX, EOS_IDX]]
                prediction = vi_sp.decode(pred_tokens)

                # Add to lists
                references.append([reference])
                hypotheses.append(prediction)

            # Update sample count
            sample_count += len(src)

            # Print progress
            if sample_count % 50 == 0:
                print(f"Processed {sample_count}/{max_samples} samples")

            # Free up memory
            del src, output_ids
            torch.cuda.empty_cache()

    # Calculate BLEU
    if not hypotheses:
        return 0.0, [], []

    print(f"Calculating BLEU score on {len(hypotheses)} samples")
    bleu_score = bleu.corpus_score(hypotheses, references).score
    return bleu_score, hypotheses, references

# ===== TASK 8: TRAINING LOOP WITH CHECKPOINT SAVING =====

def train_and_evaluate(model, train_dataloader, dev_dataloader, optimizer, criterion, scheduler,
                       num_epochs=10, patience=3, grad_accumulation_steps=1,
                       clip=1.0, use_amp=False, scaler=None,
                       save_dir='models', resume_from=None):
    """Memory-optimized training with checkpoint loading/saving."""
    os.makedirs(save_dir, exist_ok=True)

    best_loss = float('inf')
    best_bleu = 0.0
    patience_counter = 0
    start_epoch = 0

    train_losses = []
    valid_losses = []
    bleu_scores = []

    # Resume from checkpoint if specified
    if resume_from and os.path.exists(resume_from):
        print(f"Loading checkpoint from {resume_from}")
        checkpoint = torch.load(resume_from, map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])
        optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
        start_epoch = checkpoint['epoch'] + 1
        best_loss = checkpoint.get('best_loss', float('inf'))
        best_bleu = checkpoint.get('best_bleu', 0.0)
        train_losses = checkpoint.get('train_losses', [])
        valid_losses = checkpoint.get('valid_losses', [])
        bleu_scores = checkpoint.get('bleu_scores', [])
        print(f"Resuming from epoch {start_epoch}")

    for epoch in range(start_epoch, num_epochs):
        start_time = time.time()

        # Train
        train_loss = train_epoch(
            model, train_dataloader, optimizer, criterion,
            grad_accumulation_steps, clip, use_amp, scaler, scheduler
        )
        train_losses.append(train_loss)

        # Evaluate
        valid_loss = evaluate(model, dev_dataloader, criterion)
        valid_losses.append(valid_loss)

        # Calculate BLEU on a small sample
        if epoch % 2 == 0 or epoch == num_epochs - 1:  # Calculate BLEU less frequently
            bleu_score, hyps, refs = calculate_bleu_sample(
                model, dev_dataloader, lo_sp, vi_sp, device, max_samples=16
            )
        else:
            bleu_score = bleu_scores[-1] if bleu_scores else 0.0

        bleu_scores.append(bleu_score)

        end_time = time.time()
        epoch_mins, epoch_secs = divmod(end_time - start_time, 60)

        # Print progress
        print(f'Epoch: {epoch+1:02} | Time: {epoch_mins}m {epoch_secs:.2f}s')
        print(f'\tTrain Loss: {train_loss:.4f} | Val. Loss: {valid_loss:.4f} | BLEU Score: {bleu_score:.2f}')

        # Check for best model by validation loss
        if valid_loss < best_loss:
            best_loss = valid_loss
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'best_loss': best_loss,
                'best_bleu': best_bleu,
                'train_losses': train_losses,
                'valid_losses': valid_losses,
                'bleu_scores': bleu_scores,
                'params': {
                    'd_model': d_model,
                    'nhead': nhead,
                    'num_encoder_layers': num_encoder_layers,
                    'num_decoder_layers': num_decoder_layers,
                    'dim_feedforward': dim_feedforward,
                    'src_vocab_size': lo_vocab_size,
                    'tgt_vocab_size': vi_vocab_size
                }
            }, f"{save_dir}/best_model_loss.pt")
            print(f'\tBest validation loss so far: {best_loss:.4f}')
            patience_counter = 0
        else:
            patience_counter += 1

        # Check for best model by BLEU score
        if bleu_score > best_bleu:
            best_bleu = bleu_score
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'best_loss': best_loss,
                'best_bleu': best_bleu,
                'train_losses': train_losses,
                'valid_losses': valid_losses,
                'bleu_scores': bleu_scores,
                'params': {
                    'd_model': d_model,
                    'nhead': nhead,
                    'num_encoder_layers': num_encoder_layers,
                    'num_decoder_layers': num_decoder_layers,
                    'dim_feedforward': dim_feedforward,
                    'src_vocab_size': lo_vocab_size,
                    'tgt_vocab_size': vi_vocab_size
                }
            }, f"{save_dir}/best_model_bleu.pt")
            print(f'\tBest BLEU score so far: {best_bleu:.2f}')

        # Always save latest checkpoint for resuming
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'best_loss': best_loss,
            'best_bleu': best_bleu,
            'train_losses': train_losses,
            'valid_losses': valid_losses,
            'bleu_scores': bleu_scores
        }, f"{save_dir}/checkpoint_latest.pt")

        # Early stopping
        if patience_counter >= patience:
            print(f'Early stopping after {epoch+1} epochs')
            break

        # Show GPU memory usage
        if torch.cuda.is_available():
            show_gpu_memory()

    # Plot learning curves
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Train')
    plt.plot(valid_losses, label='Valid')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(bleu_scores)
    plt.xlabel('Epoch')
    plt.ylabel('BLEU Score')

    plt.tight_layout()
    plt.savefig(f'{save_dir}/learning_curves.png')

    return train_losses, valid_losses, bleu_scores

# Create directory for models
os.makedirs('models_fixed', exist_ok=True)

# Train model
train_losses, valid_losses, bleu_scores = train_and_evaluate(
    model=model,
    train_dataloader=train_dataloader,
    dev_dataloader=dev_dataloader,
    optimizer=optimizer,
    criterion=criterion,
    scheduler=scheduler,
    num_epochs=15,
    patience=5,
    grad_accumulation_steps=grad_accumulation_steps,
    clip=1.0,
    use_amp=use_amp,
    scaler=scaler,
    save_dir='models_fixed',
    resume_from=None
)
```

```python

# ===== TASK 10: INFERENCE AND TESTING =====

def translate_file(model, input_file, output_file, src_sp, tgt_sp, device, batch_size=8):
    """Memory-efficient file translation."""
    model.eval()

    # Read input file
    with open(input_file, 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f]

    results = []

    # Process in small batches
    for i in range(0, len(lines), batch_size):
        batch_lines = lines[i:i+batch_size]
        batch_tokens = []

        # Tokenize
        for line in batch_lines:
            tokens = [BOS_IDX] + src_sp.encode(line, out_type=int) + [EOS_IDX]
            if len(tokens) > 256:  # Truncate if too long
                tokens = tokens[:255] + [EOS_IDX]
            batch_tokens.append(tokens)

        # Pad sequences
        max_len = max(len(seq) for seq in batch_tokens)
        padded_batch = []
        for seq in batch_tokens:
            padded_seq = seq + [PAD_IDX] * (max_len - len(seq))
            padded_batch.append(padded_seq)

        # Convert to tensor
        src = torch.tensor(padded_batch, dtype=torch.long, device=device)

        # Generate translations
        with torch.no_grad():
            output_ids = model.generate(src, max_len=256, beam_size=3)

        # Convert to text and add to results
        for j in range(len(output_ids)):
            # Convert tokens to text
            pred_tokens = output_ids[j].tolist()
            pred_tokens = [t for t in pred_tokens if t not in [PAD_IDX, BOS_IDX, EOS_IDX]]
            translation = vi_sp.decode(pred_tokens)
            results.append(translation)

        # Free memory
        del src, output_ids
        if i % 50 == 0:
            gc.collect()
            torch.cuda.empty_cache()
            print(f"Processed {i+len(batch_lines)}/{len(lines)} lines")

    # Write results
    with open(output_file, 'w', encoding='utf-8') as f:
        for line in results:
            f.write(line + '\n')

    print(f"Translation completed and saved to {output_file}")

# Load best model for inference
best_model_path = 'models_fixed/best_model_bleu.pt'
if os.path.exists(best_model_path):
    print(f"Loading best model from {best_model_path}")
    checkpoint = torch.load(best_model_path, map_location=device)

    # Create model with same parameters
    best_model = MemoryEfficientTransformer(
        src_vocab_size=checkpoint['params']['src_vocab_size'],
        tgt_vocab_size=checkpoint['params']['tgt_vocab_size'],
        d_model=checkpoint['params']['d_model'],
        nhead=checkpoint['params']['nhead'],
        num_encoder_layers=checkpoint['params']['num_encoder_layers'],
        num_decoder_layers=checkpoint['params']['num_decoder_layers'],
        dim_feedforward=checkpoint['params']['dim_feedforward']
    ).to(device)

    # Load state dict
    best_model.load_state_dict(checkpoint['model_state_dict'])
    print(f"Model loaded with BLEU score {checkpoint.get('best_bleu', 0.0):.2f}")
else:
    print("No saved model found, using the current model")
    best_model = model

# Create output directory
os.makedirs('translations', exist_ok=True)

# Translate test file with small batch size for memory efficiency
translate_file(
    model=best_model,
    input_file=test_source_file,
    output_file='translations/test_predictions.vi',
    src_sp=lo_sp,
    tgt_sp=vi_sp,
    device=device,
    batch_size=8  # Small batch size for memory efficiency
)

# Calculate final BLEU score on test set
test_bleu, test_hyps, test_refs = calculate_bleu_sample(
    best_model, test_dataloader, lo_sp, vi_sp, device, max_samples=1000
)
print(f"BLEU score on test set: {test_bleu:.2f}")
```

```python

# ===== TASK 11: SIMPLE INTERFACE FOR TRANSLATION =====

def translate_text(model, text, src_sp, tgt_sp, device):
    """Translate a single text."""
    model.eval()

    # Tokenize
    tokens = [BOS_IDX] + src_sp.encode(text, out_type=int) + [EOS_IDX]
    src = torch.tensor([tokens], dtype=torch.long, device=device)

    # Generate translation
    with torch.no_grad():
        output_ids = model.generate(src, max_len=256, beam_size=3)

    # Convert tokens to text
    pred_tokens = output_ids[0].tolist()
    pred_tokens = [t for t in pred_tokens if t not in [PAD_IDX, BOS_IDX, EOS_IDX]]
    translation = tgt_sp.decode(pred_tokens)

    return translation

# Simple demo function
def demo_translation():
    """Interactive demo for translation."""
    print("\n==== Demo Translation ====")
    print("Enter Lao text to translate to Vietnamese (or 'exit' to quit):")

    while True:
        text = input("\nLao text: ")
        if text.lower() == 'exit':
            break

        translation = translate_text(best_model, text, lo_sp, vi_sp, device)
        print(f"Vietnamese translation: {translation}")

# Run the demo if this script is executed directly
if __name__ == "__main__":
    demo_translation()
```

```python
def save_test_results_to_drive(test_bleu, hypotheses, references, source_texts=None, drive_path='/content/drive/MyDrive/NMT_Lao_Viet_Model'):
    """
    Save test results to Google Drive.

    Args:
        test_bleu (float): The BLEU score on test data
        hypotheses (list): List of model translations (strings)
        references (list): List of reference translations (list of lists)
        source_texts (list, optional): List of source texts
        drive_path (str): Path to save in Google Drive
    """
    import os
    import csv
    import json
    from datetime import datetime

    # Create directory if it doesn't exist
    os.makedirs(drive_path, exist_ok=True)

    # Current timestamp for unique filenames
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    # 1. Save summary results text file
    summary_file = os.path.join(drive_path, f"test_results_summary_{timestamp}.txt")
    with open(summary_file, 'w', encoding='utf-8') as f:
        f.write(f"Lao-Vietnamese NMT Test Results\n")
        f.write(f"Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n")
        f.write(f"BLEU Score on test set: {test_bleu:.2f}\n\n")
        f.write(f"Number of test samples: {len(hypotheses)}\n")

        # Show a few example translations
        f.write("\nExample Translations (first 5):\n")
        f.write("="*50 + "\n")
        for i in range(min(5, len(hypotheses))):
            if source_texts:
                f.write(f"Source: {source_texts[i]}\n")
            f.write(f"Reference: {references[i][0]}\n")
            f.write(f"Translation: {hypotheses[i]}\n")
            f.write("-"*50 + "\n")

    # 2. Save detailed results as CSV
    csv_file = os.path.join(drive_path, f"test_results_detailed_{timestamp}.csv")
    with open(csv_file, 'w', encoding='utf-8', newline='') as f:
        writer = csv.writer(f)
        if source_texts:
            writer.writerow(["ID", "Source", "Reference", "Translation"])
            for i in range(len(hypotheses)):
                writer.writerow([i+1, source_texts[i], references[i][0], hypotheses[i]])
        else:
            writer.writerow(["ID", "Reference", "Translation"])
            for i in range(len(hypotheses)):
                writer.writerow([i+1, references[i][0], hypotheses[i]])

    # 3. Save just the translations (for easy inspection)
    trans_file = os.path.join(drive_path, f"translations_{timestamp}.txt")
    with open(trans_file, 'w', encoding='utf-8') as f:
        for hyp in hypotheses:
            f.write(f"{hyp}\n")

    # 4. Save references (for comparison)
    ref_file = os.path.join(drive_path, f"references_{timestamp}.txt")
    with open(ref_file, 'w', encoding='utf-8') as f:
        for ref in references:
            f.write(f"{ref[0]}\n")

    # 5. Save source texts if provided
    if source_texts:
        src_file = os.path.join(drive_path, f"sources_{timestamp}.txt")
        with open(src_file, 'w', encoding='utf-8') as f:
            for src in source_texts:
                f.write(f"{src}\n")

    # 6. Save raw data for future analysis
    raw_data = {
        "bleu": test_bleu,
        "timestamp": timestamp,
        "num_samples": len(hypotheses)
    }
    json_file = os.path.join(drive_path, f"results_metadata_{timestamp}.json")
    with open(json_file, 'w', encoding='utf-8') as f:
        json.dump(raw_data, f, indent=2)

    print(f"All test results saved to {drive_path}")
    print(f"Summary file: {os.path.basename(summary_file)}")
    print(f"Detailed results: {os.path.basename(csv_file)}")
    print(f"Translations file: {os.path.basename(trans_file)}")
```
